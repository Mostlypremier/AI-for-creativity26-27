```python
import pandas as pd
import matplotlib.pyplot as plt
path = "https://media.githubusercontent.com/media/MuseumofModernArt/collection/refs/heads/main/Artworks.csv"
df = pd.read_csv(path)
```

```python
df.info()
```

```python
year = pd.to_numeric(df['Date'], errors='coerce')
year.hist(bins=50)
plt.xlabel('year')
plt.ylabel('Number of artworks')
plt.title('MoMA artwork')
```

```python
plt.scatter(df['Width (cm)'], df['Height (cm)'], alpha=0.5)
plt.xlabel('Width (cm)')
plt.ylabel('Height (cm)')
plt.title('MoMA artwork')
```

```python
decade = (pd.to_numeric(df['DateAcquired'].str[:4]) // 10 * 10)
decade.value_counts().sort_index().plot(kind='bar')
plt.xlabel('Date Acquired')
plt.ylabel('Number of artworks')
plt.title('MoMA artwork')
```
